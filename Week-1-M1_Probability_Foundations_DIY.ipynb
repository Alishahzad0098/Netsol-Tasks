## Probability Terminology Explained

### Key Concepts with Email Examples

#### Random Experiment
- **Definition**: A process that produces an outcome
- **Email example**: Classifying an email as spam/ham
- **Other examples**: Rolling a die, flipping a coin

#### Trial
- **Definition**: One execution of a random experiment
- **Email example**: Processing one email message
- **Other examples**: One coin flip, one die roll

#### Outcome
- **Definition**: The result of a trial
- **Email example**: "spam" or "not spam" classification
- **Other examples**: "heads", "tails", die face number

#### Sample Space
- **Definition**: All possible outcomes
- **Email example**: All possible email texts and classifications
- **Notation**: S = {"spam", "not spam"}

# N-gram Language Models

## Markov Assumption
The probability of a word depends only on the previous **k** words:

P(w_n | w_1...w_{n-1}) ≈ P(w_n | w_{n-k}...w_{n-1})


## Unigram Model
**Assumption**: Words are completely independent  
**Probability**:
P(w) = count(w) / total_words

**Example**: "I love you"

P("I love you") = P("I") × P("love") × P("you")


## Bigram Model
**Assumption**: Word depends on 1 previous word  
**Probability**:
P(w_n | w_{n-1}) = count(w_{n-1} w_n) / count(w_{n-1})

**Example**: "I love you"
P("I love you") = P("I") × P("love"|"I") × P("you"|"love")

## Trigram Model
**Assumption**: Word depends on 2 previous words  
**Probability**:
P(w_n | w_{n-2}, w_{n-1}) = count(w_{n-2} w_{n-1} w_n) / count(w_{n-2} w_{n-1})

**Example**: "I love you"
P("I love you") = P("I") × P("love"|"I") × P("you"|"I love")

## Comparison Table

| Model    | Dependency          | Example Probability            |
|----------|---------------------|--------------------------------|
| Unigram  | None                | P("love")                     |
| Bigram   | 1 previous word     | P("you"/"love")              |
| Trigram  | 2 previous words    | P("you"/"I love")            |

#Data Loading & Preprocessing

````python
def load_data(filepath):
    """Load text file and return as string"""
    with open(filepath, 'r', encoding='utf-8') as file:
        text = file.read()
    return text

def preprocess(text):
    """
    Preprocess text by:
    1. Converting to lowercase
    2. Removing special characters
    3. Splitting into tokens (words)
    """
    import re

    # Convert to lowercase
    text = text.lower()


    text = re.sub(r"[^a-z0-9'\s]", "", text)

    # Split into tokens (words)
    tokens = text.split()

    return tokens

# TEST
text = load_data("input.txt")
tokens = preprocess(text)
print("First 50 tokens:", tokens[:50])
````

# Model Implementation

````python
def build_ngram_model(tokens, n=2):
    """
    Build n-gram model from tokens
    Example: For n=2 (bigram), returns {"i": {"am": 5, "have": 3}}
    """
    model = {}

    # 1. Loop through tokens with sliding window of size n
    # 2. For each window, update model counts
    # Hint: Use zip(*[tokens[i:] for i in range(n)]) for sliding window
    ngrams = zip(*[tokens[i:] for i in range(n)])

    for ngram in ngrams:
        context = tuple(ngram[:-1])  # Context is the first n-1 words
        next_word = ngram[-1]        # Next word is the nth word

        if context not in model:
            model[context] = {}

        # Increment count for the next_word given the context
        model[context][next_word] = model[context].get(next_word, 0) + 1

    return model

def predict_next_word(model, context, k=4):
    """
    Predict next words given context
    Returns: list of (word, probability) tuples
    """
    # Ensure context is a tuple for dictionary lookup
    if not isinstance(context, tuple):
        context = (context,) # Convert single word context to a tuple

    if context not in model:
        return [] # No predictions if context not found

    possible_next_words = model[context]
    total_occurrences = sum(possible_next_words.values())

    if total_occurrences == 0:
        return [] # Should not happen if context is in model, but good for safety

    # Calculate probabilities and store as (word, probability) tuples
    predictions = []
    for word, count in possible_next_words.items():
        probability = count / total_occurrences
        predictions.append((word, probability))

    # Sort by probability in descending order and return top k
    predictions.sort(key=lambda x: x[1], reverse=True)
    return predictions[:k]
````

#Testing

````python
bigram_model = build_ngram_model(tokens, n=2)
guess = input("Enter a word to predict the next word: ").strip()
print("Words after '{}':".format(guess), predict_next_word(bigram_model, guess))
````

### 1. Prior Probability (P(c))
**What it represents:**  
The baseline probability of a class before examining the message content.

**Example Calculation:**  

#### In 100 emails:
spam_count = 30

ham_count = 70

P_spam = spam_count/100  # 0.30

P_ham = ham_count/100    # 0.70

### 2. Likelihood (P(w|c))

**Definition:**  
The probability of observing a specific word given a particular class (spam/ham). This measures how characteristic a word is for a certain class.


#### Spam class: 25 occurrences out of 1000 total words
P_free_spam = 25/1000  # 0.025

#### Ham class: 2 occurrences out of 3000 total words
P_free_ham = 2/3000    # ≈0.00067

### 3. Posterior Probability

**Definition:**
The posterior probability **P(c|msg)** represents:
- The probability that a given message belongs to class *c* (spam/ham)
- The key quantity we compute to make classification decisions

### Bayesian Classification

### Decision Rule
The classifier selects the class with the highest posterior probability:

Where:
- `P(c|msg)` = Posterior probability (what we want to calculate)
- `P(msg|c)` = Likelihood of the message given the class
- `P(c)` = Prior probability of the class

### Practical Example: "free prize" message

**Given Probabilities:**

### Word probabilities
P_free_spam = 0.025

P_free_ham = 0.00067

P_prize_spam = 0.015

P_prize_ham = 0.001



### Class priors
P_spam = 0.3

P_ham = 0.7

### Calculation Steps
#### Compute Spam Probability:
P_msg_spam = P_free_spam * P_prize_spam * P_spam
           = 0.025 × 0.015 × 0.3
           ≈ 0.0001125

####Compute Ham Probability:

P_msg_ham = P_free_ham * P_prize_ham * P_ham
          = 0.00067 × 0.001 × 0.7
          ≈ 0.000000469

####Comparison:
0.0001125 (Spam) > 0.000000469 (Ham)

##Laplace Smoothing
**Problem**:  
P("xyz"|Spam) = 0 if "xyz" never appeared in training spam

**Solution**:  
$$ P_{\text{smooth}}(w|c) = \frac{\text{count}(w,c) + 1}{\text{count}(c) + V} $$

*Example*:  
- V = 10,000 words  
- count("prize", Spam) = 0  
- count(Spam) = 1000  
$$ P_{\text{smooth}}("prize"|Spam) = \frac{0+1}{1000+10000} \approx 0.00009 $$



## Stop Words in Text Processing

**Definition:** Common words (e.g., "the", "a", "and") that appear frequently but carry minimal meaning.

**Why Remove Them?**
1. **Meaning Preservation:**  
   "The quick brown fox" → "quick brown fox" (keeps core meaning)
   
2. **Improved Analysis:**  
   - Without: P("the"|spam)=0.101 vs P("the"|ham)=0.098 (useless)  
   - With: P("free"|spam)=0.025 vs P("free"|ham)=0.00067 (discriminative)

3. **Efficiency:**  
   Reduces vocabulary size by 30-40% (1000 words → 600 words)

````python
def load_sms_data(filepath):
    """Load SMS spam data from file.
    Returns: (list of messages, list of labels)"""
    messages = []
    labels = []

    with open(filepath, 'r', encoding='utf-8') as f:
        for line in f:
            parts = line.strip().split('\t')
            if len(parts) == 2:
                labels.append(parts[0])
                messages.append(parts[1])
    return messages, labels

def preprocess_text(text, remove_stopwords=True):
    """Clean text with optional stopword removal"""
    import re

    # Basic cleaning
    text = text.lower()
    text = re.sub(r'[^\w\s]', '', text)  # Remove punctuation
    text = re.sub(r'\s+', ' ', text).strip()

    if remove_stopwords:
        stop_words = {
            'i', 'me', 'my', 'myself', 'we', 'our', 'ours', 'ourselves',
            'you', 'your', 'yours', 'yourself', 'yourselves', 'he', 'him',
            'his', 'himself', 'she', 'her', 'hers', 'herself', 'it', 'its',
            'itself', 'they', 'them', 'their', 'theirs', 'themselves',
            'what', 'which', 'who', 'whom', 'this', 'that', 'these',
            'those', 'am', 'is', 'are', 'was', 'were', 'be', 'been', 'being',
            'have', 'has', 'had', 'having', 'do', 'does', 'did', 'doing',
            'a', 'an', 'the', 'and', 'but', 'if', 'or', 'because', 'as',
            'until', 'while', 'of', 'at', 'by', 'for', 'with', 'about',
            'against', 'between', 'into', 'through', 'during', 'before',
            'after', 'above', 'below', 'to', 'from', 'up', 'down', 'in',
            'out', 'on', 'off', 'over', 'under', 'again', 'further',
            'then', 'once', 'here', 'there', 'when', 'where', 'why', 'how',
            'all', 'any', 'both', 'each', 'few', 'more', 'most', 'other',
            'some', 'such', 'no', 'nor', 'not', 'only', 'own', 'same',
            'so', 'than', 'too', 'very', 's', 't', 'can', 'will', 'just',
            'don', 'should', 'now'
        }

        words = text.split()
        filtered_words = [word for word in words if word not in stop_words]
        text = ' '.join(filtered_words)

    return text

# Test with stopword removal
messages, labels = load_sms_data("SMSSpamCollection")
print("Original:", messages[0])
print("Processed:", preprocess_text(messages[0]))
print("\nWithout stopwords:", preprocess_text(messages[0], remove_stopwords=True))
````

#Calculate Prior Probabilities

````python
def calculate_priors(labels):
    prior_prob = {}
    total_messages = len(labels)
    spam_count = labels.count('spam')
    ham_count = labels.count('ham')

    prior_prob['spam'] = spam_count / total_messages
    prior_prob['ham'] = ham_count / total_messages

    return prior_prob
````

#Count Words per Class

````python
from collections import defaultdict

def count_words(messages, labels, preprocess_text_func):
    word_counts = {'spam': defaultdict(int), 'ham': defaultdict(int)}
    vocab = set()
    total_spam_words = 0
    total_ham_words = 0

    for i, message in enumerate(messages):
        label = labels[i]
        processed_words = preprocess_text_func(message, remove_stopwords=True).split()
        for word in processed_words:
            word_counts[label][word] += 1
            vocab.add(word)
            if label == 'spam':
                total_spam_words += 1
            else:
                total_ham_words += 1

    return word_counts, vocab, total_spam_words, total_ham_words
````

#Compute Word Probabilities (with Smoothing)

````python
def calculate_word_probs(word_counts, vocab, total_spam_words, total_ham_words, k=1):
    word_probs = {'spam': defaultdict(float), 'ham': defaultdict(float)}
    V = len(vocab)

    # Calculate P(word|spam)
    for word in vocab:
        count_in_spam = word_counts['spam'][word]
        word_probs['spam'][word] = (count_in_spam + k) / (total_spam_words + k * V)

    # Calculate P(word|ham)
    for word in vocab:
        count_in_ham = word_counts['ham'][word]
        word_probs['ham'][word] = (count_in_ham + k) / (total_ham_words + k * V)

    return word_probs
````

#Predict Spam Probability

````python
import math

def predict(message, prior_prob, word_probs, preprocess_text_func, vocab):
    # Preprocess message and split into words
    processed_words = preprocess_text_func(message, remove_stopwords=True).split()

    # Initialize log probabilities
    log_prob_spam = math.log(prior_prob['spam'])
    log_prob_ham = math.log(prior_prob['ham'])

    # Add log likelihoods for each word
    for word in processed_words:
        # If word is in our vocabulary, use its probability
        if word in vocab:
            log_prob_spam += math.log(word_probs['spam'][word])
            log_prob_ham += math.log(word_probs['ham'][word])
        else:
            # Handle unseen words by assigning a very small probability
            # This implicitly uses the smoothing factor k for unseen words if V is based on training vocab
            # A simpler approach for unseen words not in the `word_probs` dictionary would be to add log(k / (total_words_in_class + k * V)) directly
            # Since `calculate_word_probs` already smoothed for all words in `vocab`, words not in `vocab` would get a zero probability if not handled. This is simplified by just not adding to log_prob.
            pass # For now, we'll ignore words not in vocab as their log-prob would be handled by initial smoothing constant if they were in vocab.

    # Convert log probabilities back to regular probabilities
    spam_prob_unnormalized = math.exp(log_prob_spam)
    ham_prob_unnormalized = math.exp(log_prob_ham)

    # Calculate P(spam|message)
    if (spam_prob_unnormalized + ham_prob_unnormalized) == 0:
      # Avoid division by zero if both are extremely small
      return 0.5 # Or raise an error, or return a default
    probability = spam_prob_unnormalized / (spam_prob_unnormalized + ham_prob_unnormalized)

    return probability
````
